> Note: Words written in bold represents key topics that were studied and covered during the day.

```python
import numpy as np
```

### Step 1 - Represent three data samples as a matrix:

```python
x = np.array([
    [25, 50000, 3],
    [40, 80000, 10],
    [33, 62000, 5]
])

print("X =\n", x)
print("Shape of X:", x.shape, "3 Samples, 3 featuers")
```

### Step 2 - Dot product:

```python
features = x[0]                         # [25, 50000, 3]
weights = np.array([0.1, 0.0002, 1.5])  

# By Hand:
by_hand = (features[0] * weights[0]) +  (features[1] * weights[1]) + (features[2] * weights[2])
print("By hand:", by_hand)

# Verifiy with np.dot:
verified = np.dot(features, weights)
print("np.dot:", by_hand)
```

### Step 3 - Matrix multiplication:

```python
predictions = x @ weights

print("X Shape:", x.shape)
print("Weights Shape:", weights.shape)
print("Predictions:", predictions)

for i, pred in enumerate(predictions):
    print(f"Customer {i+1}: predicted score = {pred:.2f}")
```

### Step 4 — Deliberately a shape-mismatch:

```python
bad_weights = np.array([0.1, 0.0002, 1.5, 0.9])

try:
    broken = x @ bad_weights
except ValueError as e:
    print("ValueError raised:")
    print(e)
```
