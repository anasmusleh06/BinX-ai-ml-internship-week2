**Dataset**: Medical Cost Personal Datasets (Insurance).

```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/insurance.csv")
df.head()
```

## Hands-On Lab:

#### Step 1 & 2: 

> Load "charges" column and calculate mean, median, mode, std and IQR.

```python
charges = df["charges"]

mean = charges.mean()
median = charges.median()
mode = charges.mode()
standard_deviation = charges.std()

Q1 = charges.quantile(0.25)
Q3 = charges.quantile(0.75)
IQR = Q3 - Q1

print("Mean =", mean)
print("Median =", median)
print("Standard Deviation =", standard_deviation)
print("IQR =", IQR)
```

#### Step 3:

> Which better represents a "typical" value, mean or median?

```python
print("Mean =", mean)
print("Median =", median)
print("The Difference =", mean - median)
```

#### We notice the mean (13270) is higher than median (9382), So the **median** is better to represents a 'typical' value.

#### Step 4:

> Summary the column's center and spread:

- In general, patients pay around 13270 for medical charges, but the median is only around 9382, meaning half of all patients pay less than that. 
This tells us a small group of patients with very high bills is pulling the average up. Because of this, the median gives a better picture of what a "typical" patient actually pays.
The data is also spread out a lot (the standard deviation is close to $12110), which confirms that charges vary greatly from one to another.
